```python
!pip install ultralytics
import cv2
from ultralytics import YOLO
from google.colab.patches import cv2_imshow

yolo = YOLO('yolov8n.pt')

# Assuming the file has been uploaded using the files.upload() cell
# and we are taking the first uploaded file.
if 'uploaded' in locals() and uploaded:
    uploaded_filename = next(iter(uploaded.keys()))
    image_path = f'/content/Profile pic.jpg'
else:
    # Fallback or error if no file was uploaded, for safety
    print("No file uploaded or 'uploaded' variable not found. Please upload an image.")
    image_path = None # Or raise an error

if image_path:
    results = yolo(image_path)
    for r in results:
        im_array = r.plot()     # plot a BGR numpy array of predictions
        # OpenCV loads images as BGR, but cv2_imshow expects RGB for correct color display.
        im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB)
        cv2_imshow(im_rgb)
```
