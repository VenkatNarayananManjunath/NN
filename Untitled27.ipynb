```python
from PIL import Image
import numpy as np

# ====== 1. Load the image ======
image_path = "OIP.webp"  # replace with your image file
img = Image.open(image_path).convert("RGB")  # ensure RGB format

# ====== 2. Convert image to NumPy array ======
pixel_array = np.array(img)

# ====== 3. Print image shape ======
print("Image size (Height, Width, Channels):", pixel_array.shape)

# ====== 4. Separate RGB channels ======
red_channel = pixel_array[:, :, 0]
green_channel = pixel_array[:, :, 1]
blue_channel = pixel_array[:, :, 2]

# ====== 5. Print each channel matrix ======
print("\nRed Channel Matrix:\n")
print(red_channel)

print("\nGreen Channel Matrix:\n")
print(green_channel)

print("\nBlue Channel Matrix:\n")
print(blue_channel)

# ====== 6. Access pixel value example ======
row, col = 10, 20
print(f"\nPixel at ({row},{col}) -> R:{red_channel[row,col]}, G:{green_channel[row,col]}, B:{blue_channel[row,col]}")
```

```python
from PIL import Image
import numpy as np

# ====== 1. Load the image ======
# Replace 'input.jpg' with your image file path
image_path = "OIP.webp"
img = Image.open(image_path)

# ====== 2. Convert image to NumPy array ======
pixel_array = np.array(img)

# ====== 3. Print basic details ======
print("Image size (Height, Width, Channels):", pixel_array.shape)

# ====== 4. Display pixel values ======
print("\nPixel Values:\n")
print(pixel_array)

# ====== 5. Access individual pixel (example) ======
# Example: pixel at row 10, column 20
row, col = 10, 20
print(f"\nPixel value at ({row}, {col}):", pixel_array[row, col])
```
